# Fase 2: Preparación del dataset para el árbol de decisión
---

Convertimos Churn de Yes/No a 1/0.

```python
df['Churn'] = df['Churn'].map({
    'Yes': 1,
    'No': 0
})
```

La mayoría de variables son categóricas, así que aplicamos One-Hot Encoding para que el modelo las pueda leer. Las variables booleanas no cambian; las que tienen más de 2 valores se expanden en n-1 columnas.

```python
categorical_cols = df.select_dtypes(include=['object']).columns

df_encoded = pd.get_dummies(
    df,
    columns=categorical_cols,
    drop_first=True
)
print(f"Dataset modificado. Filas: {df_encoded.shape[0]}, Columnas: {df_encoded.shape[1]}")
```

Separamos las features (X) de la etiqueta (y).

```python
# Separamos features (X) y etiqueta (y)X = df_encoded.drop(columns=['Churn'])
y = df_encoded['Churn']
```

Dividimos en 80% entrenamiento y 20% prueba.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)
```

---
# Fase 3: Entrenamiento del árbol de decisión
---

Entrenamos 4 árboles con distintas configuraciones para comparar.

Parámetros usados:
- criterion='gini': criterio de división eficiente
- max_depth=5: limita la profundidad para evitar overfitting
- min_samples_split=15: mínimo de muestras para dividir un nodo
- min_samples_leaf=10: mínimo de muestras en cada hoja

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import plot_tree

tree_gini = DecisionTreeClassifier(
    criterion='gini',
    max_depth=5,
    min_samples_split=15,
    min_samples_leaf=10,
    random_state=42
)

tree_entropy = DecisionTreeClassifier(
    criterion='entropy',
    max_depth=5,
    min_samples_split=15,
    min_samples_leaf=10,
    random_state=42
)

tree_gini_balanced = DecisionTreeClassifier(
    criterion='gini',
    class_weight='balanced',
    max_depth=5,
    min_samples_split=15,
    min_samples_leaf=10,
    random_state=42
)

tree_entropy_balanced = DecisionTreeClassifier(
    criterion='entropy',
    class_weight='balanced',
    max_depth=5,
    min_samples_split=15,
    min_samples_leaf=10,
    random_state=42
)
```

Entrenamos cada modelo con los datos de entrenamiento.

```python
tree_gini.fit(X_train, y_train)

tree_entropy.fit(X_train, y_train)

tree_gini_balanced.fit(X_train, y_train)

tree_entropy_balanced.fit(X_train, y_train)
```

---
# Fase 4: Evaluación de los modelos
---

Comparamos accuracy en train vs test para verificar que no haya overfitting.

```python
modelos = {
    "Gini": tree_gini,
    "Entropy": tree_entropy,
    "Gini Balanced": tree_gini_balanced,
    "Entropy Balanced": tree_entropy_balanced
}

for nombre, modelo in modelos.items():

    train_score = modelo.score(X_train, y_train)

    test_score = modelo.score(X_test, y_test)

    print(f"\n===== {nombre} =====")

    print(f"Accuracy Train: {train_score:.4f}")

    print(f"Accuracy Test : {test_score:.4f}")
```

Función que evalúa cada modelo e imprime sus métricas.

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

import seaborn as sns
import matplotlib.pyplot as plt

def evaluar_modelo(modelo, X_test, y_test, nombre):

    # Predicciones sobre el set de prueba    y_pred = modelo.predict(X_test)

    # Accuracy    accuracy = accuracy_score(y_test, y_pred)

    print(f"\n===== {nombre} =====")

    print(f"\nAccuracy: {accuracy:.4f}")

    # Reporte de clasificación    print("\nReporte de ClasificaciÃ³n:")

    print(classification_report(y_test, y_pred))

    # Matriz de confusión    cm = confusion_matrix(y_test, y_pred)

    # Visualización heatmap    plt.figure(figsize=(7,5))

    sns.heatmap(
        cm,
        annot=True,
        fmt='d',
        cmap='Blues',
        xticklabels=['No Churn', 'Churn'],
        yticklabels=['No Churn', 'Churn']
    )

    plt.title(f"Matriz de ConfusiÃ³n - {nombre}")

    plt.ylabel("Real")
    plt.xlabel("PredicciÃ³n")

    plt.show()
```

Evaluamos los 4 modelos.

```python
evaluar_modelo(
    tree_gini,
    X_test,
    y_test,
    "Decision Tree - Gini"
)

evaluar_modelo(
    tree_entropy,
    X_test,
    y_test,
    "Decision Tree - Entropy"
)

evaluar_modelo(
    tree_gini_balanced,
    X_test,
    y_test,
    "Decision Tree - Gini Balanced"
)

evaluar_modelo(
    tree_entropy_balanced,
    X_test,
    y_test,
    "Decision Tree - Entropy Balanced"
)
```

Tabla resumen del accuracy de cada modelo.

```python
resultados = pd.DataFrame({
    "Modelo": [
        "Gini",
        "Entropy",
        "Gini Balanced",
        "Entropy Balanced"
    ],
    "Accuracy": [
        accuracy_score(y_test, tree_gini.predict(X_test)),
        accuracy_score(y_test, tree_entropy.predict(X_test)),
        accuracy_score(y_test, tree_gini_balanced.predict(X_test)),
        accuracy_score(y_test, tree_entropy_balanced.predict(X_test))
    ]
})

print(resultados)
```

Gráfico comparativo de accuracy.

```python
modelos = [
    "Gini",
    "Entropy",
    "Gini Balanced",
    "Entropy Balanced"
]

accuracies = [
    accuracy_score(y_test, tree_gini.predict(X_test)),
    accuracy_score(y_test, tree_entropy.predict(X_test)),
    accuracy_score(y_test, tree_gini_balanced.predict(X_test)),
    accuracy_score(y_test, tree_entropy_balanced.predict(X_test))
]

plt.figure(figsize=(8,5))

sns.barplot(
    x=modelos,
    y=accuracies
)

plt.title("ComparaciÃ³n de Accuracy")

plt.ylabel("Accuracy")

plt.ylim(0,1)

plt.show()
```

Visualizamos el árbol con mejor score.
