```python
import pandas as pd
import numpy as np

url_github = "https://raw.githubusercontent.com/JDOG-JuanDORtiz/Project4_IA/refs/heads/main/Customer-Churn.csv"
df = pd.read_csv(url_github)

print(f"Dataset cargado con Ã©xito. Filas: {df.shape[0]}, Columnas: {df.shape[1]}")
print(df.head())

print(df.info())
```

Limpieza inicial del dataset:
- Eliminamos customerID porque no aporta información predictiva.
- Convertimos TotalCharges a número (viene como texto por los clientes nuevos que tienen el campo vacío).
- Eliminamos las 11 filas con valores nulos (clientes con 0 meses de antigüedad).

```python
df = df.drop(columns=['customerID'])

df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

df = df.dropna()
df.drop_duplicates(inplace=True)
print(f"Dataset modificado. Filas: {df.shape[0]}, Columnas: {df.shape[1]}")
print(df.head())
```

La mayoría de variables son categóricas, así que aplicamos One-Hot Encoding para que el modelo pueda usarlas. Las variables booleanas no cambian; las que tienen más de 2 valores se expanden en n-1 columnas.

```python
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})

# Identificar columnas categóricas# Excluimos 'Churn' porque ya lo convertimos a 0/1# One-Hot Encodingcategorical_cols = df.select_dtypes(include=['object']).columns
df_encoded = pd.get_dummies(df, columns=categorical_cols, drop_first=True)
print(f"Dataset modificado. Filas: {df_encoded.shape[0]}, Columnas: {df_encoded.shape[1]}")
print(df_encoded.head())
# Exportar dataset codificadodf_encoded.to_csv('telco_churn_encoded.csv', index=False)

print("Â¡Archivo exportado con Ã©xito como 'telco_churn_encoded.csv'!")
```
