## Arquitectura del modelo (MLP - Multi-Layer Perceptron)

- **Entrada:** 30 variables
- **Capa oculta 1:** 64 neuronas, activación ReLU
- **Capa oculta 2:** 32 neuronas, activación ReLU
- **Salida:** 1 neurona, activación Sigmoid

**Por qué esta arquitectura:**
- Dos capas son suficientes para capturar patrones no lineales en datos tabulares.
- La forma de embudo (64 → 32) obliga a la red a destilar la información más relevante.
- ReLU evita el problema del gradiente que desaparece.
- Sigmoid en la salida da una probabilidad entre 0 y 1.

**Regularización:**
- Dropout (0.3 y 0.2): desactiva neuronas aleatoriamente durante el entrenamiento para evitar overfitting.
- Early Stopping: detiene el entrenamiento cuando la pérdida de validación deja de mejorar.

**Parámetros de entrenamiento:**
- Optimizador: Adam (ajusta el learning rate dinámicamente).
- Batch size: 32.
- Loss: Binary Cross-Entropy.

## Metodología de entrenamiento

Usamos Mini-Batch Gradient Descent con batches de 32 registros y un máximo de 100 épocas, pero con Early Stopping que detiene el entrenamiento en el punto óptimo.

Las curvas de pérdida y accuracy nos sirven para diagnosticar overfitting: si la curva de validación diverge de la de entrenamiento, el modelo está memorizando en lugar de generalizar.

```python
import matplotlib.pyplot as plt
from tensorflow.keras import callbacks

# Callback para imprimir el progreso cada 10 épocasclass LogProgress(callbacks.Callback):
    def on_epoch_end(self, epoch, logs=None):
        # Imprimimos cada 10 épocas        if (epoch + 1) % 10 == 0:
            print(f"Ã‰poca {epoch+1}: PÃ©rdida={logs['loss']:.4f}, PrecisiÃ³n={logs['accuracy']:.4f}, Val_Loss={logs['val_loss']:.4f}")

# Early Stopping: para cuando la pérdida de validación deje de mejorarearly_stop = callbacks.EarlyStopping(
    monitor='val_loss',
    patience=10,
    restore_best_weights=True,
    verbose=1 # Esto imprimirÃ¡ cuÃ¡ndo y por quÃ© se detuvo el entrenamiento
)

# Compilamos el modelomodel.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Entrenamos el modeloprint("Iniciando entrenamiento...")
history = model.fit(
    X_train_scaled, y_train,
    epochs=100,
    batch_size=32,
    validation_data=(X_test_scaled, y_test),
    callbacks=[early_stop, LogProgress()],
    verbose=1
)

# Resumen del entrenamientobest_epoch = early_stop.stopped_epoch - early_stop.patience if early_stop.stopped_epoch != 0 else 100
print(f"\nEntrenamiento finalizado. El modelo se detuvo en la Ã©poca {early_stop.stopped_epoch} (o terminÃ³ las 100).")
print(f"Mejor precisiÃ³n alcanzada en validaciÃ³n: {max(history.history['val_accuracy']):.4f}")
```
